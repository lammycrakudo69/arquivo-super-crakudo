```python
# escrever um jogo de adivinhação
# o usuaria irá digitar um numero para tentar adivinhar um numero secreto de um a 1 a 69
# o usuario tera 03 chances
# o jogo ira informar se o palpite do usuario e maoir ou menor
# informe se o palpite digitado esta quente, morno ou frio
```

```python
import random

secret_number = random.randint(1, 69)
attempts = 0
max_attempts = 3

print("Bem-vindo ao Jogo de Adivinhação!")
print("Estou pensando em um número entre 1 e 69.")
print(f"Você tem {max_attempts} chances para adivinhar.")

def get_temperature(guess, secret):
    difference = abs(guess - secret)
    if difference == 0:
        return "exatamente"
    elif difference <= 5:
        return "quente"
    elif difference <= 15:
        return "morno"
    else:
        return "frio"

while attempts < max_attempts:
    try:
        guess = int(input("Tente adivinhar o número: "))
        attempts += 1

        if guess < 1 or guess > 69:
            print("Por favor, digite um número entre 1 e 69.")
            continue

        if guess == secret_number:
            print(f"Parabéns! Você adivinhou o número {secret_number} em {attempts} tentativas!")
            break
        elif guess < secret_number:
            temperature = get_temperature(guess, secret_number)
            print(f"O número secreto é maior que {guess}. Você está {temperature}.")
        else:
            temperature = get_temperature(guess, secret_number)
            print(f"O número secreto é menor que {guess}. Você está {temperature}.")

        if attempts < max_attempts:
            remaining_attempts = max_attempts - attempts
            print(f"Você ainda tem {remaining_attempts} tentativas.")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if guess != secret_number:
    print(f"Suas {max_attempts} tentativas acabaram. O número secreto era {secret_number}.")
```
