```python
# 3 - numeros pares, o usuario digita um numero e o programa imprimi os numeros pares
num_str = input("Digite um número: ")
try:
    num = int(num_str)
    if num < 0:
        print("Por favor, digite um número positivo.")
    else:
        print(f"Números pares até {num}:")
        for i in range(num + 1):
            if i % 2 == 0:
                print(i)
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# 4 - numeros impares, o usuario digita um numero e o programa imprimi numeross impares
num_str = input("Digite um número: ")
try:
    num = int(num_str)
    if num < 0:
        print("Por favor, digite um número positivo.")
    else:
        print(f"Números ímpares até {num}:")
        for i in range(num + 1):
            if i % 2 != 0:
                print(i)
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# 5 - soma, o usuario digita um numero e o programa soma os numeros do zero ate o numero
num_str = input("Digite um número: ")
try:
    num = int(num_str)
    if num < 0:
        print("Por favor, digite um número positivo.")
    else:
        soma = 0
        for i in range(num + 1):
            soma += i
        print(f"A soma dos números de 0 até {num} é: {soma}")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# 6 - tabuada, o usuario digita um numero e o programa imprimi a tabuada do numero

num_str = input("Digite um número para a tabuada: ")
try:
    num = int(num_str)
    print(f"Tabuada de {num}:")
    for i in range(1, 11):
        print(f"{num} x {i} = {num * i}")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# 7 - contar letras, o usuario digita um numero e o programa conta as letras da palavra

word = input("Digite uma palavra: ")
num_letters = len(word)
print(f"A palavra '{word}' tem {num_letters} letras.")
```

```python
# 8 - contar vogais, o usuario digita um numero e o programa conta as vogais

word = input("Digite uma palavra: ")
vowels = "aeiouAEIOU"
vowel_count = 0
for char in word:
    if char in vowels:
        vowel_count += 1
print(f"A palavra '{word}' tem {vowel_count} vogais.")
```

```python
# 9 - maior, o usuario digita quatro numeros e o programa imprimi o maior numero digitado

numbers = []
for i in range(4):
    while True:
        try:
            num_str = input(f"Digite o {i+1}º número: ")
            num = int(num_str)
            numbers.append(num)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")

if numbers:
    max_number = numbers[0]
    for number in numbers:
        if number > max_number:
            max_number = number
    print(f"O maior número digitado é: {max_number}")
else:
    print("Nenhum número foi digitado.")
```

```python
# 10 - o usuario digita um numero entre 1 e 100 e o programa mostra quais sao divisiveis por 3 e 5

while True:
    num_str = input("Digite um número entre 1 e 100: ")
    try:
        num = int(num_str)
        if 1 <= num <= 100:
            break
        else:
            print("Número fora do intervalo. Por favor, digite um número entre 1 e 100.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

print(f"Números divisíveis por 3 e 5 até {num}:")
found_any = False
for i in range(1, num + 1):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
        found_any = True

if not found_any:
    print(f"Não há números divisíveis por 3 e 5 até {num}.")
```
