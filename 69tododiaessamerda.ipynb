```python
#Solicite ao usuário um número inteiro positivo Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

num = int(input("Por favor, digite um número inteiro positivo: "))
sum_numbers = 0

if num > 0:
    for i in range(1, num + 1):
        sum_numbers += i
    print(f"A soma dos números de 1 até {num} é: {sum_numbers}")
else:
    print("Por favor, digite um número inteiro positivo.")
```

```python
#Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

num = int(input("Por favor, digite um número inteiro positivo para calcular o fatorial: "))

if num < 0:
    print("Fatorial não existe para números negativos.")
elif num == 0:
    print("O fatorial de 0 é 1.")
else:
    fatorial = 1
    for i in range(1, num + 1):
        fatorial *= i
    print(f"O fatorial de {num} é {fatorial}.")
```

```python
#Um caixa eletrônico possui notas de:
#R$ 100, R$ 50, R$ 20, R$ 10, R$ 5, R$ 2, R$ 1

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

valor_saque = int(input("Digite o valor do saque: R$"))

notas = [100, 50, 20, 10, 5, 2, 1]
quantidade_notas = {}

print(f"Para sacar R${valor_saque}, você receberá:")

for nota in notas:
    if valor_saque >= nota:
        quantidade = valor_saque // nota
        quantidade_notas[nota] = quantidade
        valor_saque %= nota

    if quantidade_notas.get(nota, 0) > 0:
        print(f"{quantidade_notas[nota]} nota(s) de R${nota}")

if valor_saque > 0:
    print(f"Sobrou um valor de R${valor_saque} que não pode ser distribuído com as notas disponíveis.")
```
